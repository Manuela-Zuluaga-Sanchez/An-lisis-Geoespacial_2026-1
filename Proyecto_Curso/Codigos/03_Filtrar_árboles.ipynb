```python
# Cargar el archivo
ruta = r"C:/Users/manuz/OneDrive/Maestria/Análisis geoespacial/Proyecto_Curso/Datos/Datos_crudos/SAU/Arboles_SAU.gpkg"
gdf = gpd.read_file(ruta)

print(f" Total de árboles en el SAU: {len(gdf)}")

# Filtro de árboles activos en Medellín
arboles_medellin_activos = gdf[
    (gdf['municipio'] == 'MEDELLÍN') &
    (gdf['s_estado_individuo'] == 'Activo')
]

print(f" Árboles activos en Medellín: {len(arboles_medellin_activos)}")

# Guardar el archivo
ruta_salida = r"C:/Users/manuz/OneDrive/Maestria/Análisis geoespacial/Proyecto_Curso/Datos/Datos_procesados/Arboles_Activos_Medellin_TODOS.gpkg"
arboles_medellin_activos.to_file(ruta_salida, driver="GPKG")

print(f" Archivo guardado en: {ruta_salida}")
```
